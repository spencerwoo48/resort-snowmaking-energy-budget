```python
path = "C:/Users/Spencer Woo/Desktop/01_SNOW_Snowshoe_West_Virginia_Snowfall_Data/Daily_Snow_Observations_Data_2024"
```

```python
files = glob.glob(path +"/*.csv")
```

```python
li = []
```

```python
##read_csv has an argument usecols=[0,1] I can specify a selection of columns from the csv
```

```python
##each file contains data from a different month each month has a different name and not all months have the same number of days.
##I need to locate by index
##Because I haven't specified an ending index, python will read to the last column in the row regardless of how many days there are in the month
```

```python
for file in files:
    df = pd.read_csv(file, index_col=None, header=0, comment="#")##, on_bad_lines="skip")
    li.append(df.iloc[103, 7:])
```

```python
frame = pd.concat(li, axis=0)##, ignore_index=True)
```
