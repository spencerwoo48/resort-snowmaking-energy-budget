```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from bs4 import BeautifulSoup
from selenium import webdriver
from webdriver_manager.chrome import ChromeDriverManager
import time
import pandas as pd

# 1. Setup ChromeDriver
#service = Service('C:/Users/Spencer Woo/Desktop/chromedriver-win32/chromedriver-win32/chromedriver.exe') # Update path
driver = webdriver.Chrome(ChromeDriverManager().install())

# 2. URL to scrape (e.g., historical daily data)
url = 'https://www.wunderground.com/hourly/us/wv/snowshoe/date/2026-02-24'
driver.get(url)
time.sleep(5) # Give time for JS to load the table

# 3. Get page content
soup = BeautifulSoup(driver.page_source, 'html.parser')
driver.quit()

# 4. Parse the table
# Find the specific table (you may need to inspect the page to find the correct tag)
table = soup.find('hourly-forecast-table') 
df = pd.read_html(str(table))[0] # Convert table to DataFrame

print(df)
```
