```python
import pandas as pd
```

```python
import glob
```

```python
path = "C:/Users/Spencer Woo/Desktop/01_SNOW_Snowshoe_West_Virginia_Snowfall_Data/Daily_Snow_Observations_Data_2024"
```

```python
all_files = glob.glob(path + "/*.csv")
```

```python
li = []
```

```python
##read_csv has lots of arguments to direct how the data is read into the dataframe, ex delimiter and date format.
```

```python
for filename in all_files:
    df = pd.read_csv(filename, index_col=None, header=0, comment="#", on_bad_lines='skip')
    li.append(df[df["Station Name"]=="SNOWSHOE"])
```

```python
# df will only hold the most recent csv to be read into the dataframe because it is being used in the for loop
# I have to concatenate all rows in li to receive the complete dataset
```

```python
frame = pd.concat(li, axis=0, ignore_index=True)
```

```python
frame
```

```python
frame.drop_duplicates(inplace=True)
```

```python
frame
```

```python
frame.groupby('GHCN ID')
```

```python
pd.reset_option('display.max_rows')
```

```python
frame = pd.melt(frame, id_vars=['GHCN ID'], var_name='Date', value_name='Station')
```

```python
frame.to_csv('C:/Users/Spencer Woo/Desktop/01_SNOW_Snowshoe_West_Virginia_Snowfall_Data/pocahontas_county_WV_snowfall_2024.csv', index=False)
```
